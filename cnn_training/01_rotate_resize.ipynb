```python
folders = [folder for folder in os.listdir(original_dataset_path) if os.path.isdir(os.path.join(original_dataset_path, folder))]
```

```python
for folder in folders:
    folder_path = os.path.join(original_dataset_path, folder)
    output_folder_path = os.path.join(output_path, folder + '_resized')
    
    if not os.path.exists(output_folder_path):
        os.makedirs(output_folder_path)
    
    images = [image for image in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, image))]
    
    for image_name in images:
        image_path = os.path.join(folder_path, image_name)
        output_image_path = os.path.join(output_folder_path, image_name)
        
        image = Image.open(image_path)
        
        if image.width > image.height:
            
            image = image.rotate(90, expand=True)
        
        image = image.resize((250, 354))
        
        image.save(output_image_path)
        
        print(f"Processed image: {image_name}")
    
    print(f"Processed folder: {folder}")

print("Preprocessing complete")
```
