```python
import os
import random
import shutil
from PIL import Image
import numpy as np
```

```python
root_path = 'data/page_images_resized'

save_path = 'data/page_images_augmented'
```

```python
augment_folders = {
    '간지및표지_resized': 1,
    '공내재하시험_resized': 10,
    '공내전단시험_resized': 8,
    '다짐시험_resized': 20,
    '밀도시험_resized': 20,
    '본문_resized': 1,
    '사진첩_resized': 1,
    '시추공영상촬영_resized': 16,
    '시추주상도_resized': 1,
    '시험요약페이지_resized': 10,
    '실내토질시험_resized': 2,
    '암석삼축압축강도시험_resized': 20,
    '암석일축압축강도시험_resized': 10,
    '암석절리면전단시험_resized': 20,
    '점하중시험_resized': 20,
    '지반조사위치도_resized': 18,
    '지층단면도_resized': 3,
    '직접전단시험_resized': 20,
    '콘관입시험_resized': 20,
    '토질삼축압축시험_resized': 20,
    '토질일축압축시험_resized': 20,
    '표준압밀시험_resized': 20,
    '하향식탄성파시험_resized': 4,
    '하향식탄성파시험_본문_resized': 4,
    '현장베인시험_resized': 20,
    '현장수압시험_resized': 19,
    '현장투수시험_resized': 5
}
```

```python
def augment_image(image_path, save_path, times=1):
    
    if not os.path.exists(os.path.dirname(save_path)):
        os.makedirs(os.path.dirname(save_path))
        
    shutil.copy2(image_path, save_path)
    
    if times > 1:
        
        for i in range(times-1):
            img = Image.open(image_path)
            x_shift = random.randint(-15, 15)
            y_shift = random.randint(-15, 15)
            
            img_data = np.array(img)
            
            padded_img_data = np.pad(img_data, [(20, 20), (20, 20), (0, 0)], mode='constant', constant_values=255)
            
            shifted_img_data = np.roll(padded_img_data, shift=[y_shift, x_shift, 0], axis=[0, 1, 2])
            
            shifted_img_data = shifted_img_data[20:-20, 20:-20, :]
            
            shifted_img = Image.fromarray(shifted_img_data)
            
            shifted_img.save(f"{save_path[:-4]}_augmented_{i+1}.png")
```

```python
def process_folder(src_path, dst_path, augment_times):
    for filename in os.listdir(src_path):
        if filename.endswith(".png"):
            image_path = os.path.join(src_path, filename)
            save_image_path = os.path.join(dst_path, filename)
            augment_image(image_path, save_image_path, augment_times)
```

```python
for folder, times in augment_folders.items():
    folder_path = os.path.join(root_path, folder)
    save_folder_path = os.path.join(save_path, folder)
    process_folder(folder_path, save_folder_path, times)
```
